```python
num = int(input("Enter a number: "))

factorial = 1

for i in range(1, num+1):
  factorial *= i

print("Factorial of ",num ,"= ", factorial)
```

```python
num = int(input("Enter a number: "))

reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed number =", reverse)
```

```python
def is_prime(num):
    if num <= 1:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True


number = int(input("Enter a number: "))

if is_prime(number):
    print("Prime")
else:
    print("Not Prime")
```
